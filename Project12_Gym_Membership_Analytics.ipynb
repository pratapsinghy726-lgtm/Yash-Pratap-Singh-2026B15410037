## Task 2: Check missing values and duplicate `Checkin_ID` entries

```python
print("Missing values per column:")
print(df.isna().sum())
print()
dup_mask = df["Checkin_ID"].duplicated(keep=False)
print("Rows involved in duplicate Checkin_ID:", dup_mask.sum())
print("Extra (repeated) rows that will be removed:", df["Checkin_ID"].duplicated().sum())
df[dup_mask].sort_values("Checkin_ID").head(10)
```

```python
# Are the duplicates exact copies of the whole row?
print("Fully identical duplicate rows:", df.duplicated().sum())
```

Duplicates are exact copies of the same check-in (same ID, member, time and duration), so they are data-entry/system repeats. They are removed (keeping the first occurrence) so that visits are not double counted.

```python
df = df.drop_duplicates(subset="Checkin_ID", keep="first").reset_index(drop=True)
print("Rows after removing duplicates:", len(df))
print("Duplicate Checkin_IDs remaining:", df["Checkin_ID"].duplicated().sum())
```

## Task 3: Clean missing `Workout_Duration_Mins` values
**Method chosen: median imputation within each membership tier.**

**Why:**
- The median is not pulled by extreme values; the data has very long and very short sessions (see Task 13), so the mean would be distorted.
- Dropping the rows would lose valid information (the check-in itself still counts toward visits, peak hours and class attendance).
- Using the tier-wise median keeps the fill value realistic for each group of members.

```python
missing_before = df["Workout_Duration_Mins"].isna().sum()
df["Workout_Duration_Mins"] = df["Workout_Duration_Mins"].fillna(
    df.groupby("Membership_Tier")["Workout_Duration_Mins"].transform("median")
)
print("Missing before:", missing_before)
print("Missing after :", df["Workout_Duration_Mins"].isna().sum())
```

## Task 4: Convert `Checkin_Time` to datetime; extract weekday and hour

```python
df["Checkin_Time"] = pd.to_datetime(df["Checkin_Time"])
df["Weekday"] = df["Checkin_Time"].dt.day_name()
df["Hour"] = df["Checkin_Time"].dt.hour
df["Month"] = df["Checkin_Time"].dt.to_period("M").astype(str)
print(df["Checkin_Time"].dtype)
df[["Checkin_ID", "Checkin_Time", "Weekday", "Hour", "Month"]].head()
```

## Task 5: Total check-ins and average workout duration

```python
total_checkins = len(df)
avg_duration = df["Workout_Duration_Mins"].mean()
print("Total check-ins          :", total_checkins)
print("Unique active members    :", df["Member_ID"].nunique())
print("Average workout duration :", round(avg_duration, 2), "minutes")
```

## Task 6: Compare visit frequency across membership tiers

```python
tier_order = ["Basic", "Gold", "Platinum"]
n_months = df["Month"].nunique()

tier_summary = df.groupby("Membership_Tier").agg(
    Total_Checkins=("Checkin_ID", "count"),
    Unique_Members=("Member_ID", "nunique"),
    Avg_Duration_Mins=("Workout_Duration_Mins", "mean"),
).reindex(tier_order)
tier_summary["Visits_Per_Member"] = tier_summary["Total_Checkins"] / tier_summary["Unique_Members"]
tier_summary["Visits_Per_Member_Per_Month"] = tier_summary["Visits_Per_Member"] / n_months
tier_summary = tier_summary.round(2)
tier_summary
```

## Task 7: Peak gym hours and busiest days

```python
hourly = df.groupby("Hour").size().rename("Checkins").sort_values(ascending=False)
print("Top 5 peak hours:")
print(hourly.head(5))
```

```python
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
daily = df.groupby("Weekday").size().reindex(day_order).rename("Checkins")
print("Check-ins by day of week:")
print(daily)
print()
print("Busiest day :", daily.idxmax(), "(", daily.max(), "check-ins )")
print("Quietest day:", daily.idxmin(), "(", daily.min(), "check-ins )")
```

```python
# Peak-hour schedule summary: weekday x hour table of check-ins (top 3 hours per day)
heat = df.pivot_table(index="Weekday", columns="Hour", values="Checkin_ID", aggfunc="count", fill_value=0).reindex(day_order)
peak_schedule = pd.DataFrame({
    "Peak_Hour_1": heat.apply(lambda r: r.nlargest(3).index[0], axis=1),
    "Peak_Hour_2": heat.apply(lambda r: r.nlargest(3).index[1], axis=1),
    "Peak_Hour_3": heat.apply(lambda r: r.nlargest(3).index[2], axis=1),
    "Total_Checkins": heat.sum(axis=1),
})
print("PEAK-HOUR SCHEDULE SUMMARY (hour in 24h format)")
peak_schedule
```

## Task 8: Rank fitness classes by total attendance

```python
class_table = (df.groupby("Class_Name")
                 .agg(Total_Attendance=("Checkin_ID", "count"),
                      Avg_Duration_Mins=("Workout_Duration_Mins", "mean"))
                 .sort_values("Total_Attendance", ascending=False))
class_table["Rank"] = np.arange(1, len(class_table) + 1)
class_table["Share_%"] = (class_table["Total_Attendance"] / class_table["Total_Attendance"].sum() * 100)
class_table = class_table[["Rank", "Total_Attendance", "Share_%", "Avg_Duration_Mins"]].round(2)
print("CLASS POPULARITY TABLE (includes 'No Class' = general gym floor workout)")
class_table
```

```python
only_classes = class_table.drop(index="No Class", errors="ignore").copy()
only_classes["Rank"] = np.arange(1, len(only_classes) + 1)
print("Ranking of actual fitness classes only:")
only_classes
```

## Task 9: Average workout duration: male vs female

```python
gender_cmp = df.groupby("Gender")["Workout_Duration_Mins"].agg(["count", "mean", "median"]).round(2)
gender_cmp
```

```python
diff = gender_cmp.loc["Male", "mean"] - gender_cmp.loc["Female", "mean"]
print(f"Male members work out {abs(diff):.2f} minutes {'longer' if diff > 0 else 'shorter'} than female members on average.")
```

## Task 10: Monthly membership revenue by tier
**Assumption:** a member pays their tier's `Monthly_Fee` in every month in which they were active (had at least one check-in). Revenue = number of active members in that tier x monthly fee.

```python
active = df.groupby(["Month", "Membership_Tier"])["Member_ID"].nunique().rename("Active_Members").reset_index()
fee_map = df.groupby("Membership_Tier")["Monthly_Fee"].first()
active["Monthly_Fee"] = active["Membership_Tier"].map(fee_map)
active["Revenue"] = active["Active_Members"] * active["Monthly_Fee"]

revenue = active.pivot(index="Month", columns="Membership_Tier", values="Revenue")[tier_order]
revenue["Total"] = revenue.sum(axis=1)
print("Monthly revenue by tier (INR)")
revenue
```

```python
print("Total revenue over the period (INR):")
print(revenue.sum())
print()
print("Share of revenue by tier (%):")
print((revenue[tier_order].sum() / revenue["Total"].sum() * 100).round(2))
```

## Task 11: Low-engagement members (fewer than 2 visits per month)
**Definition:** average monthly visits = total check-ins of the member / number of months in the dataset. A member with an average below 2 is *low-engagement*. (Members with no check-ins at all do not appear in a check-in log, so they are not counted here.)

```python
visits = df.groupby("Member_ID").agg(
    Total_Visits=("Checkin_ID", "count"),
    Tier=("Membership_Tier", "first"),
).reset_index()
visits["Avg_Visits_Per_Month"] = visits["Total_Visits"] / n_months
low_engagement = visits[visits["Avg_Visits_Per_Month"] < 2]

print("Months in dataset                :", n_months)
print("Active members in dataset        :", len(visits))
print("LOW-ENGAGEMENT MEMBER COUNT      :", len(low_engagement))
print("Share of active members          :", round(len(low_engagement) / len(visits) * 100, 2), "%")
print()
print("Low-engagement members by tier:")
low_engagement["Tier"].value_counts().reindex(tier_order, fill_value=0)
```

## Task 12: NumPy statistics of workout duration (median, variance, quartiles)

```python
dur = df["Workout_Duration_Mins"].to_numpy()

median = np.median(dur)
variance = np.var(dur, ddof=1)      # sample variance
std_dev = np.std(dur, ddof=1)
q1, q2, q3 = np.percentile(dur, [25, 50, 75])

print("Median            :", round(median, 2))
print("Variance (sample) :", round(variance, 2))
print("Std deviation     :", round(std_dev, 2))
print("Q1 (25th pct)     :", round(q1, 2))
print("Q2 (50th pct)     :", round(q2, 2))
print("Q3 (75th pct)     :", round(q3, 2))
```

## Task 13: Unusually long or short sessions using the IQR method

```python
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print("IQR         :", round(iqr, 2))
print("Lower fence :", round(lower_fence, 2))
print("Upper fence :", round(upper_fence, 2))

too_short = df[df["Workout_Duration_Mins"] < lower_fence]
too_long = df[df["Workout_Duration_Mins"] > upper_fence]
print()
print("Unusually SHORT sessions:", len(too_short))
print("Unusually LONG sessions :", len(too_long))
```

```python
outliers = pd.concat([too_short, too_long]).sort_values("Workout_Duration_Mins")
outliers[["Checkin_ID", "Member_ID", "Membership_Tier", "Class_Name", "Workout_Duration_Mins"]].head(15)
```

Outliers are **flagged, not deleted**. Very short sessions may be accidental scans or equipment issues; very long ones may be missed check-outs. They should be reviewed by gym staff before any further action.

## Task 14: Workout-intensity label
**Documented thresholds (based on duration in minutes):**

| Label | Rule |
|---|---|
| Short | duration < 30 |
| Moderate | 30 <= duration <= 60 |
| Extended | duration > 60 |

The cut-offs follow common session lengths (a quick workout under half an hour, a standard hour-long session, and anything beyond an hour).

```python
conditions = [
    df["Workout_Duration_Mins"] < 30,
    df["Workout_Duration_Mins"].between(30, 60),
    df["Workout_Duration_Mins"] > 60,
]
labels = ["Short", "Moderate", "Extended"]
df["Intensity"] = np.select(conditions, labels, default="Unknown")

intensity_counts = df["Intensity"].value_counts().reindex(labels)
intensity_pct = (intensity_counts / len(df) * 100).round(2)
pd.DataFrame({"Sessions": intensity_counts, "Percent": intensity_pct})
```

```python
pd.crosstab(df["Membership_Tier"], df["Intensity"])[labels].reindex(tier_order)
```
